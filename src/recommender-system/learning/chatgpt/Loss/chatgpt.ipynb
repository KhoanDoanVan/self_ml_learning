# Trong Recommender System (đặc biệt là learning-to-rank), loss thường được chia 3 nhóm chính

- $\text{Pointwise}$

- $\text{Pairwise}$

- $\text{Listwise}$

## Pointwise Loss

Xem mỗi (user, item) là một instance độc lập.

Ta học:
$$
\hat{r}_{u,i} \approx r_{u,i}
$$

Ví dụ loss:
- MSE
- Binary cross-entropy

Dạng tổng quát:
$$
L = \sum_{(u,i)}{l(\hat{r}_{u,i}, r_{u,i})}
$$

Ưu điểm:
- Đơn giản
- Dễ optimize
- Phù hợp explicit rating

Nhược điểm:
- Không tối ưu ranking trực tiếp
- Không phù hợp implicit sparse data

## Pairwise Loss

Học thứ tự:
$$
\hat{x}_{u,i} > \hat{x}_{u,j}
$$

với:
- $i$ = positive
- $j$ = negative

Ví dụ:
- BPR loss
- Hinge loss

Dạng tổng quát:
$$
L = \sum_{(u,i,j)}{l(\hat{x}_{u,i} - \hat{x}_{u,j})}
$$

Ưu điểm:
- Tối ưu ranking trực tiếp
- Phù hợp implicit feedback
- Tối ưu gần với AUC

Nhược điểm:
- Cần negative sampling
- Không tối ưu toàn bộ ranking list

## Listwise Loss

Xem toàn bộ danh sách item của một user như một distribution.

Thay vì so từng cặp, ta tối ưu:
$$
P(\text{Ranking đúng})
$$

Ví dụ:
- ListNet
- ListMLE
- Softmax cross-entropy trên toàn bộ list

Dạng softmax phổ biến:

$$
P(i|u) = \frac{e^{\hat{x}_{u,i}}}{\sum_{j}{e^{\hat{x}_{u,j}}}}
$$

Loss:

$$
- \sum_{i}{y_{u,i}\log{P(i|u)}}
$$

Ưu điểm:
- Gần nhất với metric NDCG/MAP
- Tối ưu ranking toàn cục

Nhược điểm:
- Tính toán tốn kém
- Khó scale với item lớn
- Softmax over all items rất nặng

## Khi nào dùng loại nào?

- Explicit rating → Pointwise (MSE)
- Implicit feedback → Pairwise (BPR)
- Muốn tối ưu NDCG@K → Listwise

## Insight quan trọng

Ba loại này tương ứng với ba mức độ mô hình hóa:
$$
\text{Value level} → \text{Relative order level} → \text{Distribution over permutations}
$$

Listwise là mô hình hóa mạnh nhất nhưng tốn chi phí nhất.
