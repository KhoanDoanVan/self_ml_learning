# Xác định $3$ tham số bắt buộc

### 1. $\text{Size Model}$
- 7B / 13B / 34B

### 2. $\text{Avg context per request}$
- Không lấy max, lấy $P95$ (thường 512 / 1k / 2k)

### 3. $\text{Concurrent request tối thiểu}$
- ví dụ: 10 / 20 / 50

### $\rightarrow$ muốn X request → cần bao nhiêu VRAM → nên mua GPU nào

## 📎 Công thức Sizing

$$
VRAM_{total} = model_{weight} + KV_{per-token} \times avg_{tokens} \times concurrent + runtime_{buffer}
$$

Trong đó:
- $runtime_{buffer}$: thường (10-20%)

## 📎 Ví dụ

### 13B – INT4 – vLLM

#### Target: 13B model, dùng production

- Case A: Request nhẹ (chat ngắn, tool-call)
    - avg tokens: $512$
    - concurrent: $20$

$$
KV = 0.8 MB \times 512 \times 20 ≈ 8.2 GB
$$
$$
Total ≈ 8GB (model) + 8.2GB + 3GB (buffer) ≈ 19-20GB
$$

$$ \rightarrow \text{GPU tối thiểu: 24 GB} $$

- Case B: Request trung bình (chat thật)
    - avg tokens: $1k$
    - concurrent: $20$

$$
KV ≈ 16GB
$$

$$
Total ≈ 8 + 16 + 4 ≈ 28GB
$$

$$ \rightarrow \text{GPU tối thiểu: 32 GB} $$

- Case C: Muốn scale rõ ràng
    - avg tokens: $1k$
    - concurrent: $40$

$$
KV ≈ 32GB
$$

$$
Total ≈ 8 + 32 + 6 ≈ 46GB
$$

$$ \rightarrow \text{GPU tối thiểu: 48 GB trở lên} $$
