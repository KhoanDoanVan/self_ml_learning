# Calculate Error

$MSE$ (Mean Squared Error):
$$
E = E(p,y) = \frac{1}{2}.(p-y)^2
$$

- $p$: prediction
- $y$: expected outcome

# Calculate Gradient of Error Function

$$
\frac{\partial E}{\partial W^{(n)}}
$$

Imaginary, $F$ is the resulst of the neural network
$$
F = F(x,W)
$$

$\rightarrow$
$$
E = E(p,y) = E(F(x,W),y)
$$

![](image2.png)

### Calculate deriviate of $W^{(2)}:$

$$
\frac{\partial E}{\partial W^{(2)}} = \frac{\partial E}{\partial a^{(3)}}.\frac{\partial a^{(3)}}{\partial h^{(3)}}.\frac{\partial h^{(3)}}{\partial W^{(2)}}
$$

- $E = \frac{1}{2}.(a^{(3)}-y)^2 \rightarrow \frac{\partial E}{\partial a^{(3)}} = 2.\frac{1}{2}.(a^{(3)}-y).1 = a^{(3)} - y$
- $a^{(3)} = \frac{1}{1 + e^{-h^{(3)}}} \rightarrow \frac{\partial a^{(3)}}{\partial h^{(3)}} = \sigma^{'}(h^{(3)}) = \sigma{(h^{(3)})}.(1-\sigma(h^{(3)}))$
- $h^{(3)} = a^{(2)}.W^{(2)} \rightarrow \frac{\partial h^{(3)}}{\partial W^{(2)}} = a^{(2)}$

$\rightarrow$
$$
\frac{\partial E}{\partial W^{(2)}} = (a^{(3)}-y).\sigma^{'}(h^{(3)}).a^{(2)}
$$

---
### Calculate deriviate of $W^{(1)}:$
$$
\frac{\partial E}{\partial W^{(1)}} = \frac{\partial E}{\partial a^{(2)}}.\frac{\partial a^{(2)}}{\partial h^{(2)}}.\frac{\partial h^{(2)}}{\partial W^{(1)}}
$$

- $\frac{\partial E}{\partial a^{(2)}} = \frac{\partial E}{\partial a^{(3)}}.\frac{\partial a^{(3)}}{\partial h^{(3)}}.\frac{\partial h^{(3)}}{\partial a^{(2)}}$

$\rightarrow$
$$
\frac{\partial E}{\partial W^{(1)}} = \frac{\partial E}{\partial a^{(3)}}.\frac{\partial a^{(3)}}{\partial h^{(3)}}.\frac{\partial h^{(3)}}{\partial a^{(2)}}.\frac{\partial a^{(2)}}{\partial h^{(2)}}.\frac{\partial h^{(2)}}{\partial W^{(1)}}
$$

- $h^{(3)} = a^{(2)}.W^{(2)} \rightarrow \frac{\partial h^{(3)}}{\partial a^{(2)}} = W^{(2)}$
- $a^{(2)} = \frac{1}{1+e^{-h^{(2)}}} \rightarrow \frac{\partial a^{(2)}}{\partial h^{(2)}} = \sigma^{'}(h^{(2)}) = \sigma(h^{(2)})(1-\sigma(h^{(2)}))$
- $h^{(2)} = xW^{(1)} \rightarrow \frac{\partial h^{(2)}}{\partial W^{(1)}} = x$

$\rightarrow$
$$
\frac{\partial E}{\partial W^{(1)}} = (a^{(3)}-y).\sigma^{'}(h^{(3)}).W^{(2)}.\sigma^{'}(h^{(2)}).x
$$

# Backpropagation

![](image3.png)

Flow of Backpropagation:

$$
E(p,y) \rightarrow \frac{\partial E}{\partial W^{(2)}} \rightarrow \frac{\partial E}{\partial W^{(1)}}
$$

# Gradient Descent

- Take a step in opposite direction to gradient
- Step = Learning rate

![](image4.png)

- Loss is functionality of entire weight vectors:
$$
L = L(W_{1},W_{2},...,W_{n})
$$
- Global minimum or local minimum are a point in dimensions, not only any weight.
- At that point, gradient toward all dimensions equals 0:
$$
\frac{\partial L}{\partial w_{1}} = 0,...,\frac{\partial L}{\partial w_{n}}=0
$$
