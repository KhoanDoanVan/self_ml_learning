# Forward NHWC

We assume:
- Input Tensor:
$$
x \in \R^{N \times H \times W \times C}
$$

- Weights:
$$
W \in \R^{O \times C \times K \times K}
$$

- Bias:
$$
b \in \R^{O}
$$

- Stride = $s$

- Padding = $p$

## 1. Output Dimensions

$$
H_{out} = [\frac{H + 2p - K}{s}] + 1
$$
$$
W_{out} = [\frac{H + 2p - K}{s}] + 1
$$

So the output tensor is:
$$
y \in \R^{N \times H_{out} \times W_{out} \times O}
$$

## 2. Padding Operation

Define padded input:
$$
x^{pad} \in \R^{N \times (H + 2p) \times (W + 2p) \times C}
$$

With:
$$
x^{pad}_{n,h.w.c} = 
\begin{cases}
x_{n,h-p,w-p,c} & \text{if} \space p \leq h < H + p, p \leq w < W + p \\
0 & otherwise
\end{cases}
$$

## 3. Core Convolution Equation (NHWC Form)

For each:
- batch index $n$
- output spatial location $(i,j)$
- output channel $o$

$$
y_{n,i,j,o} = b_o + \sum_{c=0}^{C-1} \sum_{k_h=0}^{K-1} \sum_{k_w=0}^{K-1} {x^{pad}_{n,is+k_h,js+k_w,c}.W_{o,c,k_h,k_w}}
$$

in code:
- $h = is + k_h$ $\leftarrow$ h_start = i * stride
- $w = js + k_w$ $\leftarrow$ w_start = j * stride
- $\sum_{c,k_h,k_w} = (C \times K \times K) = \sum_{c=0}^{C-1} \sum_{k_h=0}^{K-1} \sum_{k_w=0}^{K-1}$
