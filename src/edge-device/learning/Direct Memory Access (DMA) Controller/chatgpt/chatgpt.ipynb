# Edge AI + Direct Memory Access (DMA)

### Edge AI học DMA để tránh cho CPU chết vì copy data. 
### Không hiểu DMA → không bao giờ tối ưu được latency, power, throughput.

## Edge AI bottleneck KHÔNG nằm ở model

Trên edge (mobile, embedded):
- Model thường nhỏ
- Compute không phải bottleneck chính
- Memory movement mới là kẻ giết performance

$\rightarrow$ Moving data is more expensive than computing.

## Edge AI pipeline thực tế (camera → inference)

Camera sensor
  
  ↓ (DMA)

RAM buffer
  
  ↓

NPU / GPU / DSP

  ↓ (DMA)

RAM output

  ↓

CPU post-process

### Nếu KHÔNG có DMA?

Camera → CPU register → RAM | RAM → CPU → NPU

$\rightarrow$ CPU nghẽn, pin tụt, FPS rớt.

## DMA giúp Edge AI cái gì CỤ THỂ?

1. Giảm latency end-to-end
- Không copy qua CPU
- Không cache pollution
- Pipeline song song

$\rightarrow$ Real-time (30–60 FPS) chỉ khả thi nhờ DMA.

2. Giảm power consumption
- CPU idle nhiều hơn
- DMA + accelerator energy-efficient hơn

3. Tối ưu accelerator (NPU / DSP)
- Accelerator:
    - Không đọc file
    - Không xử lý syscall
    - Chỉ đọc RAM

$\rightarrow$ DMA = cầu nối duy nhất.

4. Zero-copy & shared buffer
- Camera buffer → inference buffer
- Không memcpy
- Không allocate/free

$\rightarrow$ Đây là kỹ thuật sống còn cho edge.

## DMA trong các use case Edge AI

### Camera AI
- ISP → RAM (DMA)
- RAM → NPU (DMA)

### Audio AI
- Mic → circular buffer (DMA)
- Streaming inference

## DMA + SIMD / NEON / NPU

- DMA load tensor aligned
- NEON xử lý vector
- DMA store output

DMA (I/O) + SIMD (compute) = performance thực

Không có DMA:
- NEON nhanh mấy cũng vô nghĩa

## Học DMA ở mức nào là đủ cho Edge AI?

- DMA vs memcpy
- Cache coherence
- Buffer alignment
- Double buffering
- Interrupt vs polling
