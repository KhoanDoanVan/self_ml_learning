# Lý giải lý do của Transpose trong Neural Network

Ví dụ:

$$
z^{(1)} =
\begin{bmatrix}
z_{1}^{(1)}\\z_{2}^{(1)}\\z_{3}^{(1)}
\end{bmatrix}
=
\begin{bmatrix}
a_{1}^{(0)}.w_{11}^{(1)} + a_{2}^{(0)}.w_{21}^{(1)} + b_{1}^{(1)}\\
a_{1}^{(0)}.w_{12}^{(1)} + a_{2}^{(0)}.w_{22}^{(1)} + b_{2}^{(1)}\\
a_{1}^{(0)}.w_{13}^{(1)} + a_{2}^{(0)}.w_{23}^{(1)} + b_{3}^{(1)}\\
\end{bmatrix} = (W^{(1)})^{T}.a^{(0)} + b^{(1)}
$$

$$
a^{(1)} = \sigma(z^{(1)})
$$

Input (layer 0):
$$
a^{(0)} =
\begin{bmatrix}
a_{1}^{(0)}\\
a_{2}^{(0)}
\end{bmatrix} (2 \times 1)
$$

Layer 1 có 3 neuron $\rightarrow$ tức là có 3 outputs:

$$
z^{(1)} =
\begin{bmatrix}
z^{(1)}_{1}\\
z^{(1)}_{2}\\
z^{(1)}_{3}\\
\end{bmatrix} (3 \times 1)
$$

Mỗi neuron nhận 2 inputs $\rightarrow$ mỗi neuron có 2 weights + 1 bias

Từng neuron thực hiện:

$$
z_{1}^{(1)} = a_{1}^{(0)}.w_{11} + a_{2}^{(0)}.w_{21} + b_{1}
$$

$$
z_{2}^{(1)} = a_{1}^{(0)}.w_{12} + a_{2}^{(0)}.w_{22} + b_{2}
$$

$$
z_{3}^{(1)} = a_{1}^{(0)}.w_{13} + a_{2}^{(0)}.w_{23} + b_{3}
$$

Bây giờ trả lời câu quan trọng:
- Tại sao lại viết thành $W^{T}.a^{(0)}+b$ ?
$\rightarrow$ Để tính cùng lúc 3 neurons, ta gom weights thành 1 ma trận.

Nếu weights lưu dạng mỗi neuron là một CỘT:
- Điều này chuẩn:
$$
W^{(1)} =
\begin{bmatrix}
w_{11}&w_{12}&w_{13}\\
w_{21}&w_{22}&w_{23}
\end{bmatrix} (2 \times 3)
$$

- Giải thích:
    - 2 hàng = 2 input
    - 3 cột = 3 neuron

Vấn đề là ta muốn nhân:
- $a^{(0)}$ là (2.1)
- $W^{(1)}$ là (2.3)
$\rightarrow$ không thể nhân 2 matrixs: (2.3).(2.1) là không hợp lệ.

Nếu TRANSPOSE lại:
$$
(W^{(1)})^{T} =
\begin{bmatrix}
w_{11}&w_{21}\\
w_{12}&w_{22}\\
w_{13}&w_{23}
\end{bmatrix} (3 \times 2)
$$

Tính bằng ma trận:
$$
z^{(1)} = (W^{(1)})^{T}.a^{(0)} + b^{(1)}
$$

$\rightarrow$
$$
\begin{bmatrix}
z_{1}\\z_{2}\\z_{3}
\end{bmatrix}=
\begin{bmatrix}
w_{11}&w_{21}\\
w_{12}&w_{22}\\
w_{13}&w_{23}
\end{bmatrix}.
\begin{bmatrix}
a_{1}\\a_{2}
\end{bmatrix}+
\begin{bmatrix}
b_{1}\\b_{2}\\b_{3}
\end{bmatrix}
$$
