# Norms (Chuẩn)

Trong không gian một chiều, việc đo khoảng cách giữa hai điểm đã rất quen thuộc: lấy trị tuyệt đối của hiệu giữa hai giá trị đó.

### Trong không gian hai chiều, tức mặt phẳng, chúng ta thường dùng khoảng cách Euclid để đo khoảng cách giữa hai điểm.

Khoảng cách này chính là cái chúng ta thường nói bằng ngôn ngữ thông thường là đường chim bay. Đôi khi, để đi từ một điểm này tới một điểm kia, con người chúng ta không thể đi bằng đường chim bay được mà còn phụ thuộc vào việc đường đi nối giữa hai điểm có dạng như thế nào nữa.

### Việc đo khoảng cách giữa hai điểm dữ liệu nhiều chiều, tức hai vector, là rất cần thiết trong Machine Learning.

### Chúng ta cần đánh giá xem điểm nào là điểm gần nhất của một điểm khác; chúng ta cũng cần đánh giá xem độ chính xác của việc ước lượng; và trong rất nhiều ví dụ khác nữa.

Và đó chính là lý do mà khái niệm norm (chuẩn) ra đời. Có nhiều loại norm khác nhau mà các bạn sẽ thấy ở dưới đây:

Để xác định khoảng cách giữa hai vector $y$ và $z$, người ta thường áp dụng một hàm số lên vector $x=y-z$. Một hàm số được dùng để đo các vector cần có một vài tính chất đặc biệt.

## Định nghĩa

Một hàm số $f()$ ánh xạ một điểm $x$ từ không gian $n$ chiều sang tập số thực một chiều được gọi là norm nếu nó thoả mãn ba điều kiện sau đây:

1. $f(x) \geq 0$. Dấu bằng xảy ra $\Leftrightarrow$ $x = 0$
2. $f(\alpha x) = |\alpha|f(x), \forall \alpha \in \mathbb{R}$
3. $f(x_{1}) + f(x_{2}) \geq f(x_{1} + x_{2}), \forall x_{1},x_{2} \in R^{n}$

- Chi tiết:
    - Điều kiện thứ nhất (1): là dễ hiểu vì khoảng cách không thể là một số âm. Hơn nữa, khoảng cách giữa 2 điểm $y$ và $z$ bằng 0 nếu và chỉ nếu hai điểm nó trùng nhau, tức $x=y-z=0$
    - Điều kiện thứ hai (2): cũng có thể được lý giải như sau. Nếu 3 điểm $y,v,z$ thẳng hàng, hơn nữa $v-y=\alpha(v-z)$ thì khoảng cách giữa $v$ và $y$ sẽ gấp |$\alpha$| lần khoảng cách giữa $v$ và $z$
    - Điều kiện thứ ba (3): chính là bất đẳng thức tam giác nếu ta coi $x_{1} = w-y$, $x_{2}=z-w$ với $w$ là một điểm bất kỳ trong không gian.

## Một số chuẩn thường dùng

Giả sử các vectors $x=[x_{1};x_{2};...;x_{n}]$, $y=[y_{1};y_{2};...;y_{n}]$

Nhận thấy rằng khoảng cách Euclid chính là một Norm, norm này thường được gọi là $\text{norm 2}$:
$$
||x||_{2} = \sqrt{x_{1}^{2} + x_{2}^{2} + ... + x_{n}^2} \space (1)
$$

Với p $\text{là một số không nhỏ hơn 1}$ bất kỳ, hầm số sau đây:

$$
||x||_{p} = (|x_{1}|^p + |x_{2}|^p + ... + |x_{n}|^p)^{\frac{1}{p}} \space (2)
$$

được chính minh thoả mãn ba điều kiện bên trên, và được gọi là $\text{norm p}$.

Nhận thấy rằng khi $p \rightarrow 0$ thì biểu thức bên trên trở thành số các phần tử khác $0$ của $x$. Hàm số (2) khi $p = 0$ được gọi là giả chuẩn (pseudo-norm) 0. Nó không phải là norm vì nó không thoả mãn điều kiện 2 và 3 của norm. Giả chuẩn này, thường được ký hiệu $||x||_{0}$, khá quan trọng trong ML vì trong nhiều bài toán, chúng ta cần có ràng buộc "sparse", tức số lượng thành phần "active" của $x$ là nhỏ.

Có một vài giá trị của $p$ thường được dùng:
- KHi $p=2$ chúng ta có norm 2 như ở trên
- Khi $p=1$ chúng ta có:

$$
||x||_{1} = |x_{1}| + |x_{2}| + |x_{3}| + ... + |x_{n}| \space (3)
$$

là tổng các trị tuyệt đối của từng phần tử của $x$. Norm 1 thường được dùng như xấp xỉ của norm 0 trong các bài toán có ràng buộc "sparse". Dưới đây là một ví dụ so sánh norm 1 và norm 2 trong không gian hai chiều:

![](image1.png)

Norm 2 (màu xanh) chính là đường thằng “chim bay” nối giữa hai vector $x$ và $y$. Khoảng cách norm 1 giữa hai điểm này (màu đỏ) có thể diễn giải như là đường đi từ $x$ tới $y$ trong một thành phố mà đường phố tạo thành hình bàn cờ. Chúng ta chỉ có cách đi dọc theo cạnh của bàn cờ mà không được đi thẳng.

- KHi $p \rightarrow ∞$, ta có norm $p$ chính là trị tuyệt đối của phần tử lớn nhất của vector đó:
$$
||x||_{∞} = max_{i=1,2,...,n}|x_{i}| \space (4)
$$

## Chuẩn của ma trận

Với một ma trận $A \in \mathbb{R}^{m \times n}$, chuẩn thường được dùng nấht là chuẩn Frobenius, ký hiệu là $||A||_{F}$ là căn bậc hai của tổng bình phương tất cả các phần tử của ma trận đó.

$$
||A||_{F} = \sqrt{\sum_{i=1}^{m} \sum_{j=1}^n{a_{ij}^2}} \space (5)
$$
