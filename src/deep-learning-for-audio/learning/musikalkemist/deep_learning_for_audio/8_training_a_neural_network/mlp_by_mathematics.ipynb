# Gradient Descent

$$
W := W - \eta.\frac{\partial L}{\partial W}
$$

# Sigmoid Activation

$$
\sigma(x) = \frac{1}{1 + e^{-x}}
$$

# Sigmoid Derivative

$$
\sigma^{'}(x) = \sigma(x)(1-\sigma(x))
$$

- range: (0,1)
- smooth, nice derivative, but easy to make vanishing gradient

# Mean Square Error (MSE)

$$
L = \frac{1}{n}\sum_{i=1}^{n}{(y_{i}-\hat{y_{i}})^2}
$$

- $y_{i}$: ground truth
- $\hat{y_{i}}$: network output

# Gradient of MSE by output

General use in backprop:
$$
\frac{\partial L}{\partial \hat{y}} = 2.(\hat{y} - y)
$$

Combine with derivative of sigmoid:
$$
δ = 2.(\hat{y} - y).\hat{y}(1-y)
$$

# Formula to Update Weight in 1 Layer

Supposed:
- $a^{(l)}$: output forward layer
- $δ^{(l+1)}$: the error to backprop

Then, gradient:
$$
\frac{\partial L}{\partial W^{(l)}} = a^{(l)}.δ^{(l+1)^{T}}
$$

Then, update weights:
$$
W^{(l)} := W^{(l)}-\eta.(a^{(l)}.δ^{(l+1)^{T}})
$$
