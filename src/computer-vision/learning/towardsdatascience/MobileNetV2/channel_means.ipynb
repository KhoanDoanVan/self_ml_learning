# Channel là gì?

Channel = số feature map song song tại cùng một spatial location

Nếu Tensor có shape ($H, W, C$):
- $H, W$: không gian ảnh
- $C$: channels

### Channel trong ảnh gốc

Ảnh RGB: $224 \times 224 \times 3$

thì:
- channel 0 $\rightarrow$ Red
- channel 1 $\rightarrow$ Green
- channel 2 $\rightarrow$ Blue

Mỗi pixel có $3$ giá trị $\rightarrow$ 3 channels

### Channel trong CNN (quan trọng nhất)

Sau khi qua convolution: $224 \times 224 \times 64$

- 64 này $không$ còn là màu sắc

Mỗi channel là:
- Một feature map học được

Ví dụ (trực giác):
- Channel 1 → cạnh ngang
- Channel 7 → cạnh dọc
- Channel 23 → texture
- Channel 41 → pattern phức tạp hơn

Channel = các “cách nhìn khác nhau” của cùng một ảnh

### Tại sao channel tăng dần trong CNN?

Ví dụ pipeline:
$$
224\times224\times3
 → Conv
112\times112\times32
 → Conv
56\times56\times64
 → Conv
28\times28\times128
$$

- Spatial $\downarrow$ ($H,W \downarrow$)
- Channel $\uparrow$

Lý do:
- Mỗi pixel đại diện cho vùng ảnh lớn hơn
- Cần nhiều feature hơn để mô tả

Đổi độ chi tiết không gian lấy độ phong phú ngữ nghĩa

### Channel vs Filter 

Filter (kernel):
- Có shape: $k \times k \times C_n$
- Một filter tạo ra 1 channel output

Số filter = số channel output

Ví dụ:

Input: 32 channels

Conv dùng 64 filters

→ Output: 64 channels

### Channel trong MobileNet

$\text{Depthwise Convolution}$:
- Mỗi channel tự xử lý
- Không trộn channel

Ví dụ:

Input: 32 channels

Depthwise conv → 32 channels

$\text{Pointwise} (1\times1 conv)$:
- Trộn channel
- $\uparrow$/$\downarrow$ channel

Ví dụ:

32 → 192 (expand)

192 → 32 (project)

Vì vậy:
- Channel quyết định khả năng biểu diễn
- $1\times1$ conv quyết định độ "thông minh"
- Depthwise conv quyết định chi phí thấp
