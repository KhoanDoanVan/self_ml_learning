# Vì sao Attention phải chia cho $\sqrt{d_k}$

## 1. Phương sai ở đây là gì?

Phương sai (variance) đo độ dao động của một biến ngẫu nhiên quanh kỳ vọng.
$$
Var(X) = E[(X - E[X])^2]
$$

Trực giác:
- Variance nhỏ $\rightarrow$ giá trị ổn định
- Variacne lớn $\rightarrow$ giá trị văng xa

## 2. Dot product trong attention thực chất là gì?

Attention score:
$$
s = q.k = \sum_{i=1}^{d_k}{q_i.k_i}
$$

Giờ giả định chuẩn (đúng với init NN):
- $q_i, k_i$ độc lập
- $E[q_i] = 0$, $E[k_i] = 0$
- $Var(q_i) = Var(k_i) = 1$

## 3. Phương sai của một tích $q_i.k_i$

Với 2 biến độc lập, $mean = 0$:
$$
Var(q_i.k_i) = E[q_i^2.k_i^2] = E[q_i^2].E[k_i^2] = 1.1 = 1
$$

$\rightarrow$ Mỗi hạng tử có variance ≈ 1

## 4. Phương sai của tổng $d_k$ hạng tử

Vì các hạng tử độc lập:

$$
Var(\sum_{i=1}^{d_k}{q_i.k_i}) = \sum_{i=1}^{d_k}{Var(q_i.k_i)} = d_k
$$

$\rightarrow $Đây là câu then chốt trong đoạn bạn hỏi

## 5. Vì sao variance lớn lại nguy hiểm?

Attention score $\rightarrow$ softmax:
$$
\alpha_i = \frac{e^{s_i}}{\sum_j{e^{s_j}}}
$$

Nếu Variance lớn:
- $s_i$ rất lớn/rất nhỏ
- $e^{s_i}$ $\rightarrow$ bão hoà
- softmax ≈ one-hot
- gradient ≈ 0

#### $\rightarrow$ Attention collapse sớm

## 6. Chia $\sqrt{d_k}$ thì xảy ra chuyện gì?

$$
\~{s} = \frac{s}{\sqrt{d_k}}
$$

Phương sai:
$$
Var(\~{s}) = \frac{1}{d_k}Var(s) = \frac{1}{d_k}.d_k = 1
$$

#### $\rightarrow$ Đưa variance về mức ổn định

## 7. Trực giác cực quan trọng (nhớ câu này)

Dot product càng nhiều chiều:

$\rightarrow$ tổng càng văng

$\rightarrow$ softmax càng gắt

$\rightarrow$ gradient càng chết

Chia $\sqrt{d_k}$ = temperature normalization

## Câu chốt:
$$
\sqrt{d_k} \space \text{xuất hiện vì variance của dot product tăng tuyến tính theo số chiều}
$$

$$
chia \sqrt{d_k} \space \text{để softmax không bị bão hoà vì gradient không chết}
$$
